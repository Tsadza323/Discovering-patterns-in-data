```python
import matplotlib.pyplot as plt
import pandas as pd

# Convert columns to numeric (in case they are currently stored as strings/objects)
df["SIEVE (MM)"] = pd.to_numeric(df["SIEVE (MM)"], errors="coerce")
df["% PASSING"] = pd.to_numeric(df["% PASSING"], errors="coerce")

# Filter out any zero or NaN values which break log scales
plot_df = df[(df["SIEVE (MM)"] > 0) & (df["% PASSING"].notna())]

plt.figure(figsize=(9, 6))

plt.plot(
    plot_df["SIEVE (MM)"],
    plot_df["% PASSING"],
    marker="o"
)

plt.xlabel("Particle Size (mm)")
plt.ylabel("Percentage Passing (%)")
plt.title("Particle Size Distribution Curve")

plt.ylim(0, 100)

# Log scale and reverse axis
plt.xscale("log")
plt.gca().invert_xaxis()

plt.grid(True, which="both", linestyle="--", alpha=0.7)

plt.show()
```
