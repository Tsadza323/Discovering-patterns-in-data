```python
import pandas as pd
import matplotlib.pyplot as plt
df= pd.read_csv("/Reference_geotherm2Dmodel_base (1).csv")
df.head()
```

```python
colums = ["pres", "temp", "x","log10PER"]
df[colums].describe()
```

```python
for col in colums:
  print("\nVariable:", col)
  print("Mean:", df[col].mean())
  print("Median:", df[col].median())
  print("Range:", df[col].max() -df[col].min())
  print("Variance:", df[col].var())
  print("Standard deviation:", df[col].std())
```

```python
print("Covariance:")
print(df["temp"].cov(df["pres"]))

print("\nCorrelation:")
print(df["temp"].corr(df["pres"]))
#covariance and correlation#
```

```python
plt.scatter(df["pres"], df["temp"])
plt.xlabel("Pressure")
plt.ylabel("Temperature")
plt.title("Pressure vs Temperature")
plt.show()
#creating a plot#
```

A positive correlation means that higher values of one varaiable tend to occur with higher values of the other. A negative correlation means that higher values of one tend to occur with lower values of the other. Correlation does not necessarily mean that one variable causes the other.
